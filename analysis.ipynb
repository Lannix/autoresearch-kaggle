# Autoresearch Experiment Analysis

Analysis of autonomous PINN hyperparameter tuning results from `results.tsv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the TSV (tab-separated, 5 columns: commit, val_mse, memory_gb, status, description)
df = pd.read_csv("results.tsv", sep="\t")
df["val_mse"] = pd.to_numeric(df["val_mse"], errors="coerce")
df["memory_gb"] = pd.to_numeric(df["memory_gb"], errors="coerce")
df["status"] = df["status"].str.strip().str.upper()

print(f"Total experiments: {len(df)}")
print(f"Columns: {list(df.columns)}")
df.head(10)
```

```python
counts = df["status"].value_counts()
print("Experiment outcomes:")
print(counts.to_string())

n_keep = counts.get("KEEP", 0)
n_discard = counts.get("DISCARD", 0)
n_crash = counts.get("CRASH", 0)
n_decided = n_keep + n_discard
if n_decided > 0:
    print(f"\nKeep rate: {n_keep}/{n_decided} = {n_keep / n_decided:.1%}")
```

```python
# Show all KEPT experiments (the improvements that stuck)
kept = df[df["status"] == "KEEP"].copy()
print(f"KEPT experiments ({len(kept)} total):\n")
for i, row in kept.iterrows():
    mse = row["val_mse"]
    desc = row["description"]
    print(f"  #{i:3d}  mse={mse:.6e}  mem={row['memory_gb']:.1f}GB  {desc}")
```

## Val MSE Over Time

Track how the best (kept) val_mse evolves as experiments progress. The running minimum shows the "frontier" -- the best result achieved so far.

```python
fig, ax = plt.subplots(figsize=(16, 8))

# Filter out crashes for plotting
valid = df[df["status"] != "CRASH"].copy()
valid = valid.reset_index(drop=True)

if len(valid) > 0:
    baseline_mse = valid.loc[0, "val_mse"]
    best_mse = valid["val_mse"].min()

    # Only plot points that aren't huge outliers (e.g., limit to 5x baseline)
    below = valid[valid["val_mse"] <= baseline_mse * 5.0]

    # Plot discarded as faint background dots
    disc = below[below["status"] == "DISCARD"]
    ax.scatter(disc.index, disc["val_mse"],
               c="#cccccc", s=12, alpha=0.5, zorder=2, label="Discarded")

    # Plot kept experiments as prominent green dots
    kept_v = below[below["status"] == "KEEP"]
    ax.scatter(kept_v.index, kept_v["val_mse"],
               c="#2ecc71", s=50, zorder=4, label="Kept", edgecolors="black", linewidths=0.5)

    # Running minimum step line
    kept_mask = valid["status"] == "KEEP"
    kept_idx = valid.index[kept_mask]
    kept_mse = valid.loc[kept_mask, "val_mse"]
    running_min = kept_mse.cummin()
    ax.step(kept_idx, running_min, where="post", color="#27ae60",
            linewidth=2, alpha=0.7, zorder=3, label="Running best")

    # Label each kept experiment with its description
    for idx, mse in zip(kept_idx, kept_mse):
        desc = str(valid.loc[idx, "description"]).strip()
        if len(desc) > 45:
            desc = desc[:42] + "..."

        ax.annotate(desc, (idx, mse),
                    textcoords="offset points",
                    xytext=(6, 6), fontsize=8.0,
                    color="#1a7a3a", alpha=0.9,
                    rotation=30, ha="left", va="bottom")

    n_total = len(df)
    n_kept = len(df[df["status"] == "KEEP"])
    ax.set_xlabel("Experiment #", fontsize=12)
    ax.set_ylabel("Validation MSE (lower is better, log scale)", fontsize=12)
    ax.set_title(f"Autoresearch Progress: {n_total} Experiments, {n_kept} Kept Improvements", fontsize=14)
    ax.legend(loc="upper right", fontsize=9)
    ax.grid(True, alpha=0.2, which="both", ls="--")

    # Log scale is vital for PDE solving losses
    ax.set_yscale("log")

    # Adjust margins dynamically based on log values
    margin_factor = 1.5
    ax.set_ylim(best_mse / margin_factor, baseline_mse * margin_factor)

plt.tight_layout()
plt.savefig("progress.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved to progress.png")
```

## Summary Statistics

```python
# Summary stats
kept = df[df["status"] == "KEEP"].copy()
if len(kept) > 0:
    baseline_mse = df.iloc[0]["val_mse"]
    best_mse = kept["val_mse"].min()
    best_row = kept.loc[kept["val_mse"].idxmin()]

    print(f"Baseline val_mse:  {baseline_mse:.6e}")
    print(f"Best val_mse:      {best_mse:.6e}")
    print(f"Total improvement: {baseline_mse - best_mse:.6e} ({(baseline_mse - best_mse) / baseline_mse * 100:.2f}%)")
    print(f"Best experiment:   {best_row['description']}")
    print()

    # How many experiments to find each improvement
    print("Cumulative effort per improvement:")
    kept_sorted = kept.reset_index()
    for i, (_, row) in enumerate(kept_sorted.iterrows()):
        desc = str(row["description"]).strip()
        print(f"  Experiment #{row['index']:3d}: mse={row['val_mse']:.6e}  {desc}")
```

## Top Hits (Kept Experiments by Improvement)

```python
# Each kept experiment's delta is measured vs the previous kept experiment's mse
# (since experiments are cumulative -- each one builds on the last kept state)
kept = df[df["status"] == "KEEP"].copy()
kept["prev_mse"] = kept["val_mse"].shift(1)
kept["delta"] = kept["prev_mse"] - kept["val_mse"]

# Drop baseline (no delta)
hits = kept.iloc[1:].copy()

# Sort by delta improvement (biggest first)
hits = hits.sort_values("delta", ascending=False)

print(f"{'Rank':>4}  {'Delta':>12}  {'MSE':>12}  Description")
print("-" * 80)
for rank, (_, row) in enumerate(hits.iterrows(), 1):
    print(f"{rank:4d}  {row['delta']:+12.6e}  {row['val_mse']:12.6e}  {row['description']}")

print(f"\n{'':>4}  {hits['delta'].sum():+12.6e}  {'':>12}  TOTAL improvement over baseline")
```
